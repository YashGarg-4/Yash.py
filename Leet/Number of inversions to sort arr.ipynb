```python
count = 0
flag = False

n = int(input("N: "))
arr = [0]*n
for i in range(n):
    arr[i] = int(input(f"{i}: "))

while flag == False:
    i = 0
    while i+1 < len(arr) and arr[i] > arr[i+1]:
        arr[i+1], arr[i] = arr[i], arr[i+1]
        count+=1
        i+=1
    for i in range(len(arr)-2):
        if arr[i] > arr[i+1]:
            flag = False
            break
        else:
            flag = True

print(f'The number of inversions to take place are: {count}')
```
