```python
n = int(input("N: "))
arr = [0]*n
for i in range(n):
    arr[i] = int(input(f"{i}: "))
   
dic = {}

for i in arr:
    if i not in dic:
        dic.setdefault(i, 1)
    else:
        dic[i]+=1
       
# dic.sorted(dic.items(), key=lambda item: item[1], reverse = True)

for i in dic:
    if dic[i] >= len(arr)//2:
        print(f'The ans is: {i}')
        print(f'It occurs {dic[i]} times')
        break
else:
    print('-1 ... There exist no such majority number')
```
