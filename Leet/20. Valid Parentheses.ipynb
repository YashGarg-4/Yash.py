```python
class Solution:
    def isValid(self, s: str) -> bool:
        stack = []
        dic = {'(' : ')', '[':']', '{':'}'}
        
        for i in s:
            if stack == [] and i not in dic:
                return False
            if i in dic:
                stack.append(i)
            else:
                if i != dic[stack[-1]]:
                    print(i, stack)
                    return False
                stack.pop(-1)
        if stack == []:
            return True
        return False
                
```
